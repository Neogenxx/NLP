```python
# Installing libraries
# !pip install nltk
from nltk.util import ngrams
from collections import Counter
import re

corpus = """
Natural language processing is a field of artificial intelligence.
Natural language processing helps computers understand human language.
Language models are used in text generation and language prediction.
Artificial intelligence supports many natural language applications.
"""
```

```python
# Tokenization
tokens = re.findall(r"\b[a-z]+\b", corpus.lower()) # re is used to split the text into words and convert them to lowercase
# Generate N-grams
unigrams = list(ngrams(tokens, 1))
bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))
# Calculate frequencies
uni_freq = Counter(unigrams)
bi_freq = Counter(bigrams)
tri_freq = Counter(trigrams)
print("Total tokens:", len(tokens))
print("\nTop 5 Unigrams:")
for gram, freq in uni_freq.most_common(5):
    print(gram, freq)
print("\nTop 5 Bigrams:")
for gram, freq in bi_freq.most_common(5):
    print(gram, freq)
print("\nTop 5 Trigrams:")
for gram, freq in tri_freq.most_common(5):
    print(gram, freq)
```
